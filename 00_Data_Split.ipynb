```python
import os
import random
import pickle
from sklearn.model_selection import train_test_split
```

```python
BASE = r"C:\Users\HP\Desktop\GAVE2_preliminary\training"

all_files = sorted(os.listdir(os.path.join(BASE, "images")))

print("Total images:", len(all_files))
print(all_files[:5])
```

```python
SEED = 42
random.seed(SEED)
```

```python
train_files, temp_files = train_test_split(
    all_files,
    test_size=0.30,
    random_state=SEED
)
```

```python
val_files, test_files = train_test_split(
    temp_files,
    test_size=0.50,
    random_state=SEED
)
```

```python
print("Train:", len(train_files))
print("Validation:", len(val_files))
print("Test:", len(test_files))
```

```python
split_dict = {
    "train": train_files,
    "val": val_files,
    "test": test_files
}

with open("dataset_split.pkl", "wb") as f:
    pickle.dump(split_dict, f)
```

```python
with open("dataset_split.pkl", "rb") as f:
    split = pickle.load(f)

print(split.keys())
print(split["train"][:3])
```
